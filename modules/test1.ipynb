```python
%%time
N = 100000000
l1 = list(range(N))
for i in range(N):
    l1[i] = l1[i] * l1[i]
```

```python
%%time
N = 100000000
l2 = list(range(N))
l2 = [x*x for x in l2]
```

```python
%%time
N = 100000000
l3 = list(range(N))
l3 = map(lambda x: x*x, l3)
```

```python
%%time
N = 100000000
l1 = np.arange(N)  # Create a NumPy array with values from 0 to N-1
l1 = l1 * l1  # Vectorized operation to square all elements
```
